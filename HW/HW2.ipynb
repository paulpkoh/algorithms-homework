Practice 1: Given an array of integers, nums, where each integer is between 1 and n (n is the size of the array), some elements appear twice and others appear
once. Return an array of all the integers that appear twice in the array. You
must return the integers in their original order.

Example:

Input: [4, 3, 2, 7, 8, 2, 3, 1]

Output: [2, 3]

Answer: Go through each value of the array once and marking down the value in its own dedicated slot. If the slots already been marked, it tells you that the number is a duplicate. By the end of the array, your answer is only the duplicates.

```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]

def duplicate(nums):
  seen_twice = [False] * len(nums)
  answer = []

  for num in nums:
    if seen_twice[num - 1]:
      answer.append(num)
    else:
      seen_twice[num - 1] = True

  return answer

print(duplicate(nums))
```

Merge Sort Example:

```python
def merge_sort(arr):

  if len(arr) <= 1:
    return(arr)

  middle = len(arr) // 2
  left = merge_sort(arr[:middle])
  right = merge_sort(arr[middle:])

  return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

print(merge_sort([4, 3, 2, 7, 8, 2, 3, 1]))
```

3) Insertion sort makes a sorted section one element at a time. It slides each new element backward until it's in the correct spot. In terms of time complexity, the worst case is O(n^2) time complexity because if the array is in reverse order, it shifts past all the sorted elements again. Best case is O(N) where the input is already sorted and just one check.
