1. Adjacency Matrix

Rows are the starting node and columns are the ending node. A 0 means there is no edge. Because the graph is directed, the matrix is not symmetric (for example, A to B has weight 10, but there is no edge from B to A).

2. Adjacency List

Each node lists the nodes it points to, along with the edge weight.
- A -> (B, 10), (C, 5)
- B -> (C, 2), (D, 1)
- C -> (B, 3), (D, 9), (E, 2)
- D -> (E, 4)
- E -> (A, 7), (D, 6)

3. Python Representation with (defaultdict)

Each line lik eg['C']['E'] = 2 represents an edge: the first key is the starting node, the second key is the ending node, and the value is the weight.

```python
from collections import defaultdict
import heapq

g = defaultdict(defaultdict)
g['A']['B'] = 10; g['A']['C'] = 5
g['B']['C'] = 2; g['B']['D'] = 1
g['C']['B'] = 3; g['C']['D'] = 9; g['C']['E'] = 2
g['D']['E'] = 4
g['E']['A'] = 7; g['E']['D'] = 6
```

```python
for start in g:
    for end, weight in g[start].items():
        print(f"{start} -> {end}: {weight}")
```
