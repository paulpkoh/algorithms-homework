1. These are all examples of "Big O" notation, which is a notation that tells you how the work grows relative to the input as an upper bound.

O(1) is when the input size does not matter and the order of operations is the same regardless.
O(log N) meaning double the input adds an extra step.
O(N) is linear time where it touches every value once.
O(N^2) meaning it grows proportional to the square of the input.

2. Binary search on a sorted array would start with the middle element. If the input matches that element, then the work is done. If it isn't it checks to see if it is greater than the desired value, it eliminates the right half and vice versa for if it is smaller than the desired value. It repeats the process with the half remaining until the value is found. If it deletes everything, then the value is not in the sorted array.
