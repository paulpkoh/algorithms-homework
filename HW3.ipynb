1. Merge Two Sorted Arrays

      Input: Two sorted arrays

      Output: A merged sorted array

```python
def merge_sorted_arrays(a, b):
  i, j = 0, 0
  merged = []

  while i < len(a) and j < len(b):
    if a[i] <= b[j]:
      merged.append(a[i])
      i += 1
    else:
      merged.append(b[j])
      j += 1

  while i < len(a):
      merged.append(a[i])
      i += 1
  while j < len(b):
      merged.append(b[j])
      j += 1

  return merged
```

Merge Sort (Iterative)

Implement merge sort without recursion.

```python
def merge_sort_iterative(arr):
  result = list(arr)
  n = len(result)
  width = 1

  while width < n:
    for left in range(0, n, 2 * width):
      mid = min(left + width, n)
      right = min(left + 2 * width, n)
      result[left:right] = merge_sorted_arrays(result[left:mid], result[mid:right])
      width *= 2

  return result
```

Merge Sort (Recursive)

Implement merge sort using recursion.

```python
def merge_sort_recursive(arr):
  if len(arr) <= 1:
    return list(arr)

  mid = len(arr) // 2
  left_sorted = merge_sort_recursive(arr[:mid])
  right_sorted = merge_sort_recursive(arr[mid:])

  return merge_sorted_arrays(left_sorted, right_sorted)
```
